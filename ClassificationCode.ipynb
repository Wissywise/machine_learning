```python
import pandas as pd
import  numpy as np
import matplotlib.pyplot as plt
import  seaborn as sns
```

```python
df = pd.read_csv("Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df.info() # The column 'TotalCharges' is supposed to be numerical but showing string
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges, errors='coerce')  # Convert the TotalCharges column to numerical value
```

```python
df.info()
```

```python
df.dropna(how='any', inplace=True) # Dropping the 11 missing rows fron the TotalCharges on all the columns for uniformity.
                                   # Please note: The missing value could be imputed using mmean, mode etc
```

```python
df.info()
```

```python
len(df)
```

```python
df.head()
```

```python
df.Churn.value_counts()/len(df)*100
```

```python
# Getting my independent variable, in this case the X variable
X=df.drop(['customerID', 'Churn'], axis=1)
```

```python
# Getting the dependent variable, in this case the y-variable
y = df['Churn']
```

```python
y = y.map({'Yes':1, 'No':0})  # Converting the categorical values to numerical values
```

```python
X.columns
```

```python
y.head()
```

```python
# Convert categorical features to numericals using Feature Encoding and Dummy Encoding
X = pd.get_dummies(X, columns=['gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod'], drop_first=True, dtype='int')
```

```python
X.head(3)
```

```python
X.info()
```

```python
# Splitting the data into training and testing set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=42)
```

```python
X_train.columns
```

```python
len(X_train), len(X_test)
```

```python
len(y_train), len(y_test)
```

```python
# Feature Scaling to ensure all the values are in range not too far from each other
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

```python
X_train_sc
```

```python
X_test_sc
```

kNN Classifier

```python
# Calling the kNN Classifier
from sklearn.neighbors import KNeighborsClassifier

#initiating the Classifier
model = KNeighborsClassifier()

# Passing the data to the Classifier
model.fit(X_train_sc, y_train)
```

```python
y_pred = model.predict(X_test_sc)
```

```python
y_pred
```

```python
y_test.head()
```

```python
# Classification metrics to check how well the model is performing
from sklearn.metrics import accuracy_score

print(accuracy_score(y_pred, y_test)*100)
```
