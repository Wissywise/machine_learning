```python
'''
RandomOverSampler is a technique used to address class imbalance in datasets by increasing the number of instances in the minority class. It works by randomly duplicating examples from the minority class until the desired balance between classes is achieved. This can help improve the performance of machine learning models, especially when dealing with imbalanced datasets.
'''
from imblearn.over_sampling import RandomOverSampler

# Upsample the minority class using RandomOverSampler
oversampler = RandomOverSampler(sampling_strategy='auto', random_state=42, shrinkage=None)
# Resample the training data to balance the classes
X_train_upsampled, y_train_upsampled = oversampler.fit_resample(X_train_sc, y_train)

# Model Building
model_dt2 = DecisionTreeClassifier()
model_dt2.fit(X_train_upsampled, y_train_upsampled)

# Predict on the test set
y_pred_dt2 = model_dt2.predict(X_test_sc)
```

```python
'''
RandomUnderSampler is a technique used to address class imbalance in datasets by reducing the number of instances in the majority class. It works by randomly selecting a subset of examples from the majority class until the desired balance between classes is achieved. This can help improve the performance of machine learning models, especially when dealing with imbalanced datasets.
'''
from imblearn.under_sampling import RandomUnderSampler

# Downsample the majority class using RandomUnderSampler
downsampler = RandomUnderSampler(sampling_strategy='auto', random_state=42, replacement=False)
# Resample the training data to balance the classes
X_train_downsampled, y_train_downsampled = downsampler.fit_resample(X_train_sc, y_train)

# Model Building
model_dt3 = DecisionTreeClassifier()
model_dt3.fit(X_train_downsampled, y_train_downsampled)

# Predict on the test set
y_pred_dt3 = model_dt3.predict(X_test_sc)
```

```python
"""
SMOTE (Synthetic Minority Over-sampling Technique) from the Imbalanced-Learn SMOTE Documentation, it creates new, artificial data points for the minority class via K-Nearest Neighbors interpolation instead of blindly copying existing rows. Finds Nearest Neighbors: It targets a minority class instance and looks at its k nearest neighbours (default is 5) within the same minority class. Generates Synthetic Samples: For each minority instance, it randomly selects one of its k nearest neighbours and generates a new synthetic instance along the line segment connecting the two instances. This process is repeated until the desired balance between classes is achieved.
"""
from imblearn.over_sampling import SMOTE
# Create a SMOTE object
smote = SMOTE(random_state=42, sampling_strategy='auto', k_neighbors=5)

# Resample the data
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_sc, y_train)

# Model Building
model_dt4 = DecisionTreeClassifier()
model_dt4.fit(X_train_resampled, y_train_resampled)

# Predict on the test set
y_pred_dt4 = model_dt4.predict(X_test_sc)
```

```python
# Calculate accuracy and Print classification report
accuracy = round(accuracy_score(y_test, y_pred_dt4) * 100, 2)
print("DT SMOTE Accuracy:", accuracy, '%')
print("DT SMOTE Classification Report:")
print(classification_report(y_test, y_pred_dt4))
```

```python
"""
SMOTEENN is a hybrid technique from the imblearn library that combines SMOTE (oversampling) with Edited Nearest Neighbors (ENN) (undersampling).
While standard SMOTE is great at creating synthetic minority points, it often creates them too close to—or directly inside—the majority class boundary. This creates a noisy, blurry decision boundary. SMOTEENN solves this by cleaning up the noise after oversampling.
1. Oversample (SMOTE): It creates synthetic data points for your minority class to balance the dataset.
2. Clean & Undersample (ENN): It examines every data point using a K-Nearest Neighbors approach. If a data point's neighbors mostly belong to the opposite class, that data point is removed. This deletes noisy majority points and poorly placed synthetic minority points.
The result is a clean, distinct "no man's land" between your two classes, making it much easier for your machine learning model to find the correct decision boundary.
"""
from imblearn.combine import SMOTEENN

# Create a SMOTEENN object
smote_enn = SMOTEENN(random_state=42, sampling_strategy='auto', smote=None, enn=None, n_jobs=1)

# Resample the data
X_train_resampled, y_train_resampled = smote_enn.fit_resample(X_train_sc, y_train)

# Model Building
model_dt5 = DecisionTreeClassifier()
model_dt5.fit(X_train_resampled, y_train_resampled)

# Predict on the test set
y_pred_dt5 = model_dt5.predict(X_test_sc)

#
```

```python
# Calculate accuracy and Print classification report
accuracy = round(accuracy_score(y_test, y_pred_dt5) * 100, 2)
print("DT SMOTEENN Accuracy:", accuracy, '%')
print("DT SMOTEENN Classification Report:")
print(classification_report(y_test, y_pred_dt5))
```

```python
"""
ADASYN (Adaptive Synthetic Sampling) is an advanced oversampling technique from the imblearn library. It builds upon standard SMOTE by shifting its focus to the most difficult-to-learn regions of your dataset.
While SMOTE generates synthetic data points uniformly across all minority examples, ADASYN adaptively creates more synthetic data in areas where the minority class is heavily outnumbered by the majority class.
ADASYN determines exactly where to generate data by looking at the density distribution:
1. Measures Local Difficulty: For every minority data point, it looks at its K-Nearest Neighbors. It calculates the ratio of majority class examples surrounding that point.
2. Assigns Weights: A minority point surrounded by many majority points is flagged as "hard to learn" and given a high weight. A minority point surrounded by other minority points is flagged as "easy to learn" and given a low weight.
3. Generates Adaptively: It generates significantly more synthetic data points around the high-weight (difficult) examples and fewer around the easy ones.
This forces the machine learning model to focus on the challenging boundary regions where it is most likely to make errors.
Critical Traps & When to Avoid ADASYN:
While ADASYN is highly effective, it has one major drawback: outliers.
If your minority class contains extreme outliers or noise (e.g., an incorrectly labeled data point deep inside the majority class territory), ADASYN will see it as the "hardest" point to learn. It will mistakenly generate a massive cluster of synthetic data right around that noisy outlier, creating false patterns.
"""
from imblearn.over_sampling import ADASYN

# Create an ADASYN object
adasyn = ADASYN(random_state=42)

# Resample the data
X_train_resampled, y_train_resampled = adasyn.fit_resample(X_train_sc, y_train)

# Model Building
model_dt6 = DecisionTreeClassifier()
model_dt6.fit(X_train_resampled, y_train_resampled)

# Predict on the test set
y_pred_dt6 = model_dt6.predict(X_test_sc)
```

```python
# Calculate accuracy and Print classification report
accuracy = round(accuracy_score(y_test, y_pred_dt6) * 100, 2)
print("DT ADASYN Accuracy:", accuracy, '%')
print("DT ADASYN Classification Report:")
print(classification_report(y_test, y_pred_dt6))
```

```python
"""
AllKNN is an under-sampling technique from the imblearn.under_sampling module. It is a smart data-cleaning tool that focuses entirely on removing noisy, ambiguous majority class data points right along the decision boundary.
Unlike standard RandomUnderSampler (which blindly deletes data at random), AllKNN uses spatial mathematics to eliminate data points that blur the line between your classes.
How AllKNN Works Behind the Scenes:
AllKNN is a strategic expansion of Edited Nearest Neighbors (ENN). It cleans the majority class in a series of loop iterations:
1. Iteration 1 (k=1): It checks the 1 nearest neighbor of every majority class data point. If that neighbor belongs to the minority class (meaning the majority point is isolated or floating in enemy territory), that majority point is deleted.
2. Iteration 2 (k=2): It expands the neighborhood size by 1. It checks the 2 nearest neighbors of the surviving points and removes those that mismatch.
3. Subsequent Iterations ((k=3, ..., n)): It continues increasing the value of k by 1 at each round.
The process stops when it reaches your maximum specified n_neighbors (the default is 3) or if the majority class shrinks so much that it accidentally becomes the minority class
"""
from imblearn.under_sampling import AllKNN

# Create an AllKNN object
allknn = AllKNN(sampling_strategy='auto', n_neighbors=3, n_jobs=1)

# Resample the data
X_train_resampled, y_train_resampled = allknn.fit_resample(X_train_sc, y_train)

# Model Building
model_dt7 = DecisionTreeClassifier()
model_dt7.fit(X_train_resampled, y_train_resampled)

# Predict on the test set
y_pred_dt7 = model_dt7.predict(X_test_sc)
```

```python
# Calculate accuracy and Print classification report
accuracy = round(accuracy_score(y_test, y_pred_dt7) * 100, 2)
print("DT ALLkNN Accuracy:", accuracy, '%')
print("DT ALLkNN Classification Report:")
print(classification_report(y_test, y_pred_dt7))
```

```python
"""
SMOTETOMEK is a powerful hybrid resampling technique from the imblearn.combine module. Just like SMOTEENN, it aims to solve the problem of blurry class boundaries by combining oversampling and undersampling into a unified, two-step workflow.
Instead of using ENN to clean up data points, it specifically hunts down and eliminates Tomek Links.
What is a Tomek Link?
A Tomek Link occurs when two data points from opposite classes are each other’s immediate nearest neighbours.
Imagine a "Yes Churn" data point and a "No Churn" data point sitting incredibly close together in your feature space, isolated from their own respective groups. This pairing creates a high level of ambiguity for your machine learning model. By identifying these links and deleting the majority class instance (or both), you widen the gap between your classes.
How SMOTETOMEK Works:
1. Oversample (SMOTE): It first uses standard SMOTE to generate synthetic minority data points, bringing the classes to an equal balance.
2. Find and Destroy (Tomek Links): It scans the newly expanded dataset to identify all Tomek Links. It then removes the majority class data point from every pair.
This leaves you with a much cleaner, more distinct separation zone between your target categories.
"""
from imblearn.combine import SMOTETomek
```
